# AGN Reverberation Mapping SURD Analysis

This notebook provides a template for applying the Synergistic–Unique–Redundant Decomposition (SURD) framework to real AGN reverberation mapping data.

## 1. Setup and Data Loading

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

# Helper functions (also provided in agn_helpers.py)
def standardize(data):
    return (data - np.mean(data)) / np.std(data)

def interpolate_light_curve(times, fluxes, new_times, kind='linear'):
    f = interp1d(times, fluxes, kind=kind, fill_value="extrapolate")
    return f(new_times)

# Load NGC 5548 AGN Watch Data
url_cont = "https://raw.githubusercontent.com/manus-ai/agn-data/main/ngc5548_agnwatch_cont_clean.csv" # Placeholder for actual upload
url_hb = "https://raw.githubusercontent.com/manus-ai/agn-data/main/ngc5548_agnwatch_hb_clean.csv"

# For now, we assume local files are uploaded to Colab
try:
    df_cont = pd.read_csv('ngc5548_agnwatch_cont_clean.csv')
    df_hb = pd.read_csv('ngc5548_agnwatch_hb_clean.csv')
    print("Data loaded successfully.")
except:
    print("Please upload the processed CSV files to the Colab environment.")
````

## 2. Pre-processing for SURD

SURD requires time-series data on a common, regularly sampled grid.

````python
dt = 1.0 # 1 day cadence
t_min = max(df_cont['mjd'].min(), df_hb['mjd'].min())
t_max = min(df_cont['mjd'].max(), df_hb['mjd'].max())
common_times = np.arange(t_min, t_max, dt)

c_flux_resampled = interpolate_light_curve(df_cont['mjd'], df_cont['flux'], common_times)
hb_flux_resampled = interpolate_light_curve(df_hb['mjd'], df_hb['flux'], common_times)

c_std = standardize(c_flux_resampled)
hb_std = standardize(hb_flux_resampled)

X = np.vstack([c_std, hb_std])
print(f"Input array X shape: {X.shape}")
````

## 3. Visualization

````python
plt.figure(figsize=(12, 5))
plt.plot(common_times, c_std, label='Standardized Continuum (5100A)')
plt.plot(common_times, hb_std, label='Standardized H-beta', alpha=0.7)
plt.xlabel('MJD')
plt.ylabel('Standardized Flux')
plt.legend()
plt.title('NGC 5548 Prepared for SURD Analysis')
plt.show()
````
